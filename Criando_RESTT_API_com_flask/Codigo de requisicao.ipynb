```python
import requests
```

```python
# Requisição de um produto que existe
link = 'http://127.0.0.1:5000/vendas/produtos/Bermuda'
requisicao = requests.get(link)
print(requisicao.json())
```

```python
# Requisição de um produto que não existe

link = 'http://127.0.0.1:5000/vendas/produtos/drone'
requisicao = requests.get(link)
print(requisicao.json())
```

#### Requisições possíveis nessa api
- http://id - faturamento
- http://id//vendas/produtos - todas as vendas de todos os produtos da base de dados
- http://id//vendas/produtos/NOME_DO_PRODUTO - Vendas de um produto específico
